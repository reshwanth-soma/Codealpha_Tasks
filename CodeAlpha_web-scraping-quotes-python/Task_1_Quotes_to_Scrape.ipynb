Requests is used to fetch web page content.
BeautifulSoup is used to parse HTML.
Pandas is used to store scraped data.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

This step sends a request to the website and retrieves the HTML content.

```python
url = "https://quotes.toscrape.com/"
response = requests.get(url)
```

BeautifulSoup parses the HTML so that elements can be extracted easily.

```python
soup = BeautifulSoup(response.text, "html.parser")
```

This step extracts quotes and author names from the website.

```python
quotes = soup.find_all("div", class_="quote")

data = []

for quote in quotes:
    text = quote.find("span", class_="text").text
    author = quote.find("small", class_="author").text
    data.append([text, author])
```

The extracted data is converted into a table format using Pandas.

```python
df = pd.DataFrame(data, columns=["Quote", "Author"])
df.head()
```

This step saves the scraped data into a CSV file, creating a custom dataset as required by the PDF.

```python
df.to_csv("quotes_dataset.csv", index=False)
```

This step downloads the csv file that is created.


```python
from google.colab import files
files.download("quotes_dataset.csv")
```

This task focuses on collecting data from a public website using web scraping techniques.
Python libraries such as Requests and BeautifulSoup were used to access and parse the website's HTML content.
Relevant information was accurately extracted and organized into a structured tabular format.
The extracted data was converted into a dataset using Pandas for easy analysis.
Finally, the dataset was saved as a CSV file for further use.
